```python
# Find how many unique areas are there
pd.unique(df["area"])
df["area"].nunique()
```

```python
# Find the average price of all houses in the dataset
df["average_price"].mean()
```

**Filtering & Sorting**

```python
# Filter all rows where average_price is greater than 500000
df[df["average_price"]>500000]
```

```python
# Find all data for "barnet" area only
df[df["area"]=="barnet"]
```

```python
# Sort the dataset by average_price in descending order
df.sort_values(by="average_price",ascending=False)
```

```python
# Find the area with the highest average price
df[df["average_price"]==df["average_price"].max()]["area"]
```

```python
# Find rows where no_of_crimes is null
df["no_of_crimes"].isnull().sum()

```

```python
# Find mean average price for each area using groupby
df.groupby("area")["average_price"].mean()
```

```python
df.columns
```

```python
# Find total houses sold per area
df.groupby("area")["houses_sold"].sum()
```

```python
# Which area has the most houses sold overall?
total=df.groupby("area")["houses_sold"].sum()
```

```python
total.idxmax()
```

```python
# Find minimum and maximum average price for each area
df.groupby("area")["average_price"].agg(["max","min"])
```
